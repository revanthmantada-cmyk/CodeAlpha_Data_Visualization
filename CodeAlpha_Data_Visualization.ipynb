```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python
df = pd.read_csv('student_data.csv')

print("Dataset loaded successfully!")

print("Dataset Shape:", df.shape)

df.head()
```

```python
print("Dataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(data=df, x='G3', bins=20, kde=True)

plt.title('Distribution of Student Final Grades')
plt.xlabel('Final Grade')
plt.ylabel('Number of Students')

plt.show()
```

```python
study_avg = df.groupby('studytime')['G3'].mean()

plt.figure(figsize=(8, 5))

sns.barplot(x=study_avg.index, y=study_avg.values)

plt.title('Study Time vs Average Final Grade')
plt.xlabel('Study Time Category')
plt.ylabel('Average Final Grade')

plt.show()
```

```python
plt.figure(figsize=(12, 8))

correlation = df.corr(numeric_only=True)

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap of Student Performance')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(y=df['absences'])

plt.title('Distribution of Student Absences')
plt.ylabel('Number of Absences')

plt.show()
```

```python
gender_counts = df['sex'].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    gender_counts,
    labels=gender_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Gender Distribution of Students')

plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.stripplot(
    data=df,
    x='studytime',
    y='G3',
    jitter=True
)

plt.title('Study Time vs Final Grade')
plt.xlabel('Study Time Category')
plt.ylabel('Final Grade')

plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Chart 1: Grade distribution
sns.histplot(data=df, x='G3', bins=20, ax=axes[0, 0])
axes[0, 0].set_title('Final Grade Distribution')

# Chart 2: Study time vs average grade
sns.barplot(
    x=study_avg.index,
    y=study_avg.values,
    ax=axes[0, 1]
)
axes[0, 1].set_title('Average Grade by Study Time')

# Chart 3: Absences
sns.boxplot(y=df['absences'], ax=axes[1, 0])
axes[1, 0].set_title('Student Absences')

# Chart 4: Gender distribution
axes[1, 1].pie(
    gender_counts,
    labels=gender_counts.index,
    autopct='%1.1f%%'
)
axes[1, 1].set_title('Gender Distribution')

plt.suptitle('Student Performance Visualization Dashboard',
             fontsize=16)

plt.tight_layout()
plt.show()
```

```python
print("STUDENT PERFORMANCE - KEY FINDINGS")

print("\n1. Total Students:", len(df))

print("2. Average Final Grade:", round(df['G3'].mean(), 2))

print("3. Highest Final Grade:", df['G3'].max())

print("4. Lowest Final Grade:", df['G3'].min())

print("5. Average Absences:", round(df['absences'].mean(), 2))

print("\n6. Average Grade by Study Time:")
print(df.groupby('studytime')['G3'].mean().round(2))

print("\n7. Gender Distribution:")
print(df['sex'].value_counts())
```
